```python
fig1 = plt.figure()

plt.loglog(n_tasks,runtime_1_holes[2], color="#20B2AA", label = '$q=1; v = 64$', linewidth=2)
plt.loglog(n_tasks,runtime_1_holes_less_vi[2], ':', color="#20B2AA", label = '$q=1; v = 40$', linewidth=2)

plt.loglog(n_tasks,runtime_5_holes[2], color="#FF6EB4", label = '$q=5; v = 64$', linewidth=2)
plt.loglog(n_tasks,runtime_5_holes_less_vi[2], ':', color="#FF6EB4", label = '$q=5; v = 40$', linewidth=2)

plt.loglog(n_tasks,runtime_6_holes[2], color="#9B59B6", label = '$q=6; v=64$', linewidth=2)
plt.loglog(n_tasks,runtime_6_holes_less_vi[2], ':', color="#9B59B6", label = '$q=6; v = 40$', linewidth=2)

plt.loglog(n_tasks,np.array([8,4,2,1]), '--', color="black", label = 'ideal strong scaling', linewidth=1.5)

ax = plt.gca()  # get current axes


plt.legend()
plt.xlabel('Nodes')
plt.ylabel('time in [min]')
plt.title('Runtime disc with $q$ holes')

# Nur an diesen Stellen Ticks anzeigen:
custom_ticks = [1, 2, 4, 8]
custom_labels = [str(tick) for tick in custom_ticks]

ycustom_ticks = [0.5, 1, 2, 4, 8]
ycustom_labels = [str(tick) for tick in ycustom_ticks]

# Setze nur diese Ticks (verhindert zusätzliche)
ax.set_xticks(custom_ticks)
ax.set_xticklabels(custom_labels)

ax.set_yticks(ycustom_ticks)
ax.set_yticklabels(ycustom_labels)

# Deaktiviere automatisch generierte Minor Ticks
ax.xaxis.set_minor_locator(plt.NullLocator())
ax.yaxis.set_minor_locator(plt.NullLocator())

plt.ylim(top=12)

plt.savefig("Runtime1.png", format='png', dpi=300, bbox_inches='tight')
plt.savefig('Runtime1.eps', format="eps")
plt.show()
```
